```python
count=1
while(count<=5):
  print(f'processing step {count}')
  count+=1
print("done")
```

```python
time_left=5
while(time_left>0):
  print(f"time reaming :{time_left}s")
  time_left-=1
print("time is up")
```

```python
num=1
total=0
while(num<=4):
  total+=num
  num+=1
total
```

```python
retry=1
while(retry<=3):
  print(f"attempt{retry}:retry connection...")
  retry+=1
else:print("Max retries reached. Connection failed.")
```
