```python
tokens = ["Deep", "Learning", "is", "THE", "future", "OF", "AI"]
stopwords = ["is", "the", "of"]
clean = []

for token in tokens:
    normalized = token.lower()
    if normalized not in stopwords:
        clean.append(normalized)

print(clean)  # ['deep', 'learning', 'future', 'ai']
```
