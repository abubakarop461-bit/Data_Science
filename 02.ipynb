```python
status_code=int(input("enter your code"))
match status_code:
  case 200:
    print("Status 200: OK / Success")
  case 400:
    print("Status 400:  bad request")
  case 404:
    print("Status 404: not found")
  case 500:
    print("Status 500: internal server error")
  case _:
    print("Status Unknown: Unhandled Code")
```

```python
token_count = 1500
match token_count:
  case x if x < 500:
    print("Small prompt - Route to Haiku")
  case x if 500 <= x <2000:
    print("Medium prompt - Route to Sonnet")
  case x if x >= 2000:
    print("Large prompt - Route to Opus / R1")
  case _:
    print("invalid token count")
```
